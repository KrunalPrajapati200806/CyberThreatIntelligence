```python
from pathlib import Path
import pandas as pd
import numpy as np

DATA_DIR = Path("../data/raw/CIC-IDS2017")

files = list(DATA_DIR.glob("*.csv"))

print(f"Total CSV files: {len(files)}")
```

```python
file_path = files[0]

print("Inspecting:")
print(file_path.name)
```

```python
df = pd.read_csv(file_path)

print("Dataset shape:", df.shape)
```

```python
print("Number of columns:", len(df.columns))

for i, column in enumerate(df.columns):
    print(i, repr(column))
```

```python
df.info()
```

```python
missing = df.isnull().sum()

missing = missing[missing > 0]

print("Columns containing missing values:")
print(missing)
```

```python
print("Total missing values:", df.isnull().sum().sum())
```

```python
numeric_df = df.select_dtypes(include=np.number)

inf_count = np.isinf(numeric_df).sum()

print("Infinite values by column:")
print(inf_count[inf_count > 0])

print("\nTotal infinite values:")
print(inf_count.sum())
```

```python
duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
duplicate_percentage = (duplicate_count / len(df)) * 100

print(f"Duplicate percentage: {duplicate_percentage:.2f}%")
```

```python
label_column = [col for col in df.columns if col.strip().lower() == "label"][0]

print("Label column:", repr(label_column))
```

```python
print(df[label_column].value_counts())
```

```python
label_distribution = df[label_column].value_counts()

label_percentage = (
    df[label_column]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print(label_percentage)
```

```python
import matplotlib.pyplot as plt

label_counts = df[label_column].value_counts()

plt.figure(figsize=(12, 6))
label_counts.plot(kind="bar")
plt.title("Attack Class Distribution")
plt.xlabel("Class")
plt.ylabel("Number of Flows")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
df.describe().T
```

```python
numeric_summary = df.select_dtypes(include=np.number).describe().T

numeric_summary["range"] = (
    numeric_summary["max"] - numeric_summary["min"]
)

display(
    numeric_summary.sort_values(
        "range",
        ascending=False
    ).head(20)
)
```

```python
unique_counts = df.nunique().sort_values()

print(unique_counts.head(20))
```

```python
print("\nHighest-cardinality columns:")
print(unique_counts.tail(20))
```

```python
for col in df.columns:
    if any(keyword in col.lower() for keyword in ["ip", "timestamp", "port"]):
        print(repr(col))
```

```python
profile = {
    "rows": len(df),
    "columns": len(df.columns),
    "missing_values": int(df.isnull().sum().sum()),
    "infinite_values": int(np.isinf(
        df.select_dtypes(include=np.number)
    ).sum().sum()),
    "duplicate_rows": int(df.duplicated().sum()),
    "classes": int(df[label_column].nunique())
}

profile
```

```python
with open("../reports/dataset_profile.txt", "w") as f:
    f.write("CIC-IDS2017 Dataset Profile\n")
    f.write("=" * 40 + "\n\n")

    for key, value in profile.items():
        f.write(f"{key}: {value}\n")

    f.write("\nClass Distribution\n")
    f.write("-" * 40 + "\n")
    f.write(df[label_column].value_counts().to_string())

print("Profile saved successfully.")
```
